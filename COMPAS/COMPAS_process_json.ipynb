```python
import csv
import json

compas_JSON = {}

with open('compas-scores-two-years.csv', newline='') as csv_file:
    csv_reader = csv.reader(csv_file, delimiter='|')
    headers = next(csv_reader)[0].split(',')
    for row in csv_reader:
        row = row[0].split(',')
        compas_JSON[row[0]] = {}
        for i in range(len(headers) - 1):
            compas_JSON[row[0]][headers[i + 1]] = row[i + 1]

dir = '/content/drive/MyDrive/Explanation_XAI/COMPAS/compas_data_processed.json'
with open(dir, 'w') as json_file:
    json.dump(compas_JSON, json_file)
```
