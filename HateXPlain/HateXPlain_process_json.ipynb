```python
import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
import json

porter = PorterStemmer()

nltk.download('stopwords')
nltk.download('wordnet')
stopword_set = set(stopwords.words('english'))
lemma = nltk.wordnet.WordNetLemmatizer()

dataset_raw = open('dataset.json')
ids_raw = open('post_id_divisions.json')
ids = json.load(ids_raw)
dataset = json.load(dataset_raw)

ids_concat = ids['test'] + ids['train'] + ids['val']
count = 0

tokens = set()
for post_id in ids_concat:
    post = dataset[post_id]
    for token in post['post_tokens']:
        if token in stopword_set:
            continue
        token_stem = porter.stem(token)
        if not token_stem in tokens:
            if count < 40:
                print(token_stem)
                count += 1
        tokens.add(token_stem)

tokens_list = list(tokens)
tokens_list.sort()
print('tokens_len: {}'.format(len(tokens)))
tokens_dir = '/content/drive/MyDrive/Explanation_XAI/HateXPlain/tokens.json'
with open(tokens_dir, 'w') as f:
    json.dump(tokens_list, f)
```
